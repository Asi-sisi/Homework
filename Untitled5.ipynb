```python
def collcet_scores():
  math_scores=[]

  print('开始输入全班数学成绩')
  print('提示：若成绩已输入完毕，请输入英文字母q来结束')

  while True:
    user_input=input('请输入数学成绩：')
    if user_input=='q':
        break

    score=int(user_input)
    math_scores.append(score)
  return math_scores

def print_scores(dl):
  print("===成绩清单===")
  for i in range(len(dl)):
    print(f'第{i+1}位学生的成绩：{dl[i]}分')

def calculate_average(dl):
  if len(dl)==0:
    print("没有成绩")
    return
  all=sum(dl)
  eeg=len(dl)
  ans=int(all / eeg)

  print("--------------------------------")
  print(f"全班总分：{all}")
  print(f"平均成绩：{ans}")

x=collcet_scores()
y=print_scores(x)
z=calculate_average(x)
```
